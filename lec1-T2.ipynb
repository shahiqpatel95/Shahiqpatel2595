# password validation 
1. lenght of password more than 7 
2. atleast one upper character
3. atleast one lower character
4. atleast one special character
5. atleast one numeric value
6. no space is allowed
   if any condition not satisfy write proper massage

```python
password = input("enter password")
if len(password) > 7 :
    up=low=sp=di=space=0
    for i in password:
        if i.isupper():
            up += 1 
        elif i.islower():
            low += 1 
        elif i.isnumeric():
            di += 1
        elif i.isspace():
            space += 1
        else :
            sp += 1
    if up == 0 :
        print("Enter atleast one upper character ")
    elif low == 0 :
        print("Enter atleat one lower character")
    elif sp == 0 :
        print("Enter atleast one special character")
    elif di == 0 :    
        print("Enter atleast one digit ")
    elif space != 0 :
        print("Space not allowed")
    else :
        print("Valid Password")
else :
    print("Required length > 7 , Enter Valid Password")
```
