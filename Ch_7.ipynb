```python
%%writefile demo1.py
def cap(s):
    return s.upper()
def small(s):
    return s.lower()
def scount(s,char):
    return s.count(char)
def sfind(s,char):
    return s.find(char)
```

```python
import demo1
s = "B7 is best shahiq is not best"
char = "best"
print(demo1.cap(s))
print(demo1.small(s))
print(demo1.scount(s,char))
print(demo1.sfind(s,char))
```
