```python
num1 = int(input('enter number :'))
digit = len(str(num1))
total = 0 
for i in range(digit) :
    r = num1%10 
    total += r 
    num1 = num1//10 
print(f"total is {total}")
```

```python
n = int(input("enter number :"))
for i in range(1,n+1):
    for j in range(1, n-i+1):
        print(" ", end=" ")
    for k in range(1,i+1):
        print("*",end=" ")
    for l in range(1,i):
        print("*",end=" ")
    print()
for i in range(1,n):
    for m in range(1,i+1):
            print(" ",end=" ")
    for t in range(1,n-i+1):
            print("*",end=" ")
    for o in range(1,n-i):
            print("*", end=" ")
    print()
```

```python
n = int(input("enter row :"))
for i in range(1,n+1):
    for j in range(1,n-i+1):
        print(" ",end=" ")
    for k in range(i,0,-1):
        print(k,end=" ")
    for l in range(2,i+1):
        print(l,end=" ")
    print()
for i in range(1,n):
    for m in range(1,i+1):
        print(" ",end=" ")
    for t in range(n-i,0,-1):
        print(t,end=" ")
    for o in range(2,n-i+1):
        print(o, end=" ")
    print()
```

```python
n = int(input("enter row :"))
for i in range(1,n+1):
    for j in range(1,n-i+1):
        print(" ",end=" ")
    for k in range(i,0,-1):
        print(chr(k+64),end=" ")
    for l in range(2,i+1):
        print(chr(l+64),end=" ")
    print()
for i in range(1,n):
    for m in range(1,i+1):
        print(" ",end=" ")
    for t in range(n-i,0,-1):
        print(chr(t+64),end=" ")
    for o in range(2,n-i+1):
        print(chr(o+64), end=" ")
    print()
```

```python
n = int(input("enter number :"))
for i in range(1,n+1):
    for s in range(1,i+1):
        print(s, end=" ")
    for h in range(0,2*(n-i)-1):
        print(" " ,end=" ")
    for a in range(i if i< n else i-1 ,0,-1):
        print(a,end=" ") 
    print()
for i in range(1, n):
    for s in range(1,n-i+1):
        print(s,end=" ")
    for h in range(0,2*i-1):
        print(" ",end=" ")
    for a in range(n-i,0,-1):
        print(a,end=" ")
    print()
```
