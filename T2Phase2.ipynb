# slicing 

- step size positive
- 1. direction l to r
  2. default start 0 end=len[1](exclude) step=1



- step size nagtive
- 1. direction r to l
  2. default start = -1 end= -len[1]-1 (exclude)

```python
l=[1,2,3,4,5,6]
print(l[::])
print(l[0::])
print(l[:len(l):])
print(l[::-1])
print(l[::-2])
print(l[::-3])
print(l[-2::-2])
print(len(l[0:1]))
print(len(l[1]))
```

```python
l=[[1,2],[3,4],[5,6]]
print(l[::-1])
print(l[0:2])
print(len(l[0]))
print(len(l[0:1]))
print(l[0:1])
```

# l=[5,6,1,2,7,8,6]

```python
l=eval(input(("enter a list")))
candy=[1]*len(l)
# l to r 
for i in range(1,len(l)):
    if l[i-1]<l[i]:
        candy[i]=candy[i-1]+1
print(candy)
for i in range(-2,-len(l)-1,-1):
    if l[i]>l[i+1]:
        candy[i]=max(candy[i],candy[i+1]+1)
print(candy)
print(sum(candy))
```
