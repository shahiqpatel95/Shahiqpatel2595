```python
def is_happy_sad(n):
    temp = n
    while temp!=1 and temp!= 4:
        total = 0 
        while temp>0 :
            reminder = temp%10 
            total = total+reminder**2
            temp = temp//10 
        temp = total 
    if temp == 1 :
        return f"{n} is happy number"
    else :
        return f"{n} is sad number"
print(is_happy_sad(7))
```
